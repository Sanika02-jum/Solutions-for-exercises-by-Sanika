```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
sns.set_style('whitegrid')
```

```python
titanic = sns.load_dataset('titanic')
```

```python
titanic.head()
```

```python
sns.jointplot(data=titanic,x="fare",y="age",kind="scatter",height=5,ratio=7,space=0.2)
```

```python
sns.displot(titanic['fare'],kde=False,color='red',height=4,aspect=1)
```

```python
sns.boxplot(data=titanic,x='class',y='age',palette='rainbow')
```

```python
sns.swarmplot(data=titanic,x='class',y='age',palette='Set2')
```

```python
sns.countplot(data=titanic,x='sex',palette=['blue','green'])
```

```python
sns.heatmap(titanic.corr(numeric_only=True),cmap='coolwarm')
```
